```python
from dataclasses import replace
from pathlib import Path

from future_rag.application.container import (
    ApplicationContainer,
)
from future_rag.config.settings import RAGConfig
from future_rag.pipelines.basic_rag import BasicRAG
from future_rag.services.rag_service import RAGService


PROJECT_ROOT = Path.cwd().parent
config = RAGConfig()

chunk_1024_config = replace(
    config,
    chunk_size=1024,
    chunk_overlap=128,
    collection_name="future_rag_chunk_1024",
)

container = ApplicationContainer(
    config=chunk_1024_config,
    project_root=PROJECT_ROOT,
)

container.initialize()

pipeline = BasicRAG(
    retriever=container.retriever,
    llm=container.llm,
)

rag_service = RAGService(
    pipeline=pipeline,
)
```

```python
answer = rag_service.query(
    "What are the six guiding principles of the FUTURE-AI framework?"
)

print(answer)
```
