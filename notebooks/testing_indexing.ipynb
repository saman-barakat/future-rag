```python
from future_rag.indexing.embeddings import EmbeddingFactory

print("=" * 60)
print("Testing EmbeddingFactory")
print("=" * 60)

embedding_model = EmbeddingFactory.create()

print(f"Model Type: {type(embedding_model).__name__}")

text = "Artificial intelligence is transforming healthcare."

embedding = embedding_model.get_text_embedding(text)

print(f"Embedding Type : {type(embedding).__name__}")
print(f"Embedding Size : {len(embedding)}")
print(f"First 10 Values: {embedding[:10]}")

print("\n✅ EmbeddingFactory test passed!")
```

```python
from future_rag.knowledge.loaders import DocumentLoader
from future_rag.knowledge.chunker import DocumentChunker

from future_rag.indexing.indexer import IndexBuilder

print("=" * 60)
print("Testing IndexBuilder")
print("=" * 60)

# Load documents
documents = DocumentLoader.load("../data/knowledge/pdf")
print(f"Loaded Documents : {len(documents)}")

# Chunk documents
chunker = DocumentChunker()
nodes = chunker.split(documents)
print(f"Generated Nodes  : {len(nodes)}")

# Build index
builder = IndexBuilder()
index = builder.build(nodes)

print(f"Index Type       : {type(index).__name__}")

print("\n✅ IndexBuilder test passed!")
```
